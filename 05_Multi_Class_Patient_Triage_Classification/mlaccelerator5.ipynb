```python
import pandas as pd
import numpy as np

# ==========================================================
# EMERGENCY PATIENT TRIAGE ANALYSIS
# ==========================================================

# TASK 1 — Dataset Construction

data = {
    'Patient_ID': ['P_501', 'P_502', 'P_503', 'P_504', 'P_505',
                   'P_506', 'P_507', 'P_508', 'P_509', 'P_510'],

    'Heart_Rate_BPM': [72, 125, 95, 68, 140, 88, 75, 130, 92, 70],

    'Systolic_BP': [120, 165, 138, 115, 180, 130, 122, 170, 135, 118],

    'Oxygen_Sat_Pct': [98, 89, 94, 99, 85, 95, 97, 88, 93, 98],

    'Age': [35, 68, 45, 28, 75, 52, 40, 62, 50, 30],

    'Triage_Category': ['General', 'ICU', 'Urgent', 'General', 'ICU',
                        'Urgent', 'General', 'ICU', 'Urgent', 'General']
}

df = pd.DataFrame(data)


# TASK 2 — Supervised Paradigm Identification

paradigm = "Multi-Class Classification"
target_type = "Discrete Categorical"


# TASK 3 — Input-Output Splitting

X = df[['Heart_Rate_BPM',
        'Systolic_BP',
        'Oxygen_Sat_Pct',
        'Age']]

y = df['Triage_Category']


# TASK 4 — Multi-Class Target Distribution

class_counts = y.value_counts()
class_percentages = y.value_counts(normalize=True) * 100


# TASK 5 — Group-Wise Feature Means

group_means = df.groupby('Triage_Category')[
    ['Heart_Rate_BPM',
     'Systolic_BP',
     'Oxygen_Sat_Pct']
].mean()

group_means = group_means.reindex(['General', 'Urgent', 'ICU'])


# ==========================================================
# EXPECTED OUTPUT
# ==========================================================

print("=" * 10, "EMERGENCY PATIENT TRIAGE ANALYSIS", "=" * 10)

print("\nDataset Shape              :", df.shape[0], "Rows,", df.shape[1], "Columns")

print("\nSupervised Paradigm        :", paradigm,
      "(", target_type, "Target)")

print("\nFeatures (X)               :",
      "Heart_Rate_BPM, Systolic_BP, Oxygen_Sat_Pct, Age")

print("Target (y)                 : Triage_Category")

print("\nClass Distribution:")

for category in ['General', 'Urgent', 'ICU']:
    print(f"- {category:<10}: "
          f"{class_counts[category]} "
          f"({class_percentages[category]:.1f}%)")


print("\nTriage Level Group Averages:")

print(group_means.round(2))


print("\nAlgorithm Guidance:")
print("Decision Trees or Multi-Class Logistic Regression")
print("are recommended for interpretable clinical rule boundaries.")
```

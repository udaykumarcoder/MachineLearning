```python
import pandas as pd
```

```python
data=pd.read_csv('/kaggle/input/datasets/udaykumarmethre/churndataset/churn.csv')
print(data)
```

```python
print("Number of Rows    :", data.shape[0])
print("Number of Columns :", data.shape[1])
```

```python
features = [
    "Account_Age_Days",
    "Monthly_Spend",
    "Support_Calls",
    "Inactivity_Days"
]

target = "Churn"

print("Features:")
for feature in features:
    print(feature)

print("\nTarget:")
print(target)
```

```python
churn_counts = data["Churn"].value_counts()

print("Churn Status Counts:")
print(churn_counts.to_string())
```

```python
churned_avg=data[data["Churn"]=="Yes"]["Inactivity_Days"].mean()
not_churned_avg=data[data["Churn"]=="No"]["Inactivity_Days"].mean()
print(f"Average Inactivity Days:")
print(f"Churned Customers   : {churned_avg} days")
print(f"Retained Customers : {not_churned_avg} days")
```

```python
data["Prediction"] = data.apply(
    lambda row: "Yes"
    if row["Inactivity_Days"] > 30 and row["Monthly_Spend"] < 50
    else "No",
    axis=1
)

```

```python
# print(data.columns)
data.columns=data.columns.str.strip()
print(data[["Customer_ID", "Churn", "Prediction"]].rename(
    columns={"Churn": "Actual"}
).to_string(index=False
           ))
```

```python
cp=(data["Actual"]==data["Prediction"]).sum()
tp=len(data["Prediction"])
acc=int((cp/tp)*100)
print(f"Correct Predictions : {cp}")
print(f"Total Predictions   : {tp}")
print(f"Rule-Based Accuracy : {acc}%")
```

```python
customer_A={
    "Inactivity_Days":  32, "Monthly_Spend" : 150.0, "Support_Calls" : 6
}
customer_B={
    "Inactivity_Days": 28, "Monthly_Spend" : 30.0, "Support_Calls" : 8
}
def pred_customer(customer):
    if customer["Inactivity_Days"]>30 and customer["Monthly_Spend"]<50:
        return "Yes"
    else:
        return "No"
prediction_a=pred_customer(customer_A)
prediction_b=pred_customer(customer_B)

print(f"Customer A Prediction : {prediction_a}")
print(f"Customer B Prediction : {prediction_b}")
```

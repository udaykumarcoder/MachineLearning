```python
import pandas as pd
```

```python
data = {
    "Employee_ID": ["E101", "E102", "E103", "E104", "E105"],
    "Distance_KM": [5, 18, 7, 22, 10],
    "Travel_Time_Min": [20, 65, 30, 75, 40],
    "Previous_Late_Count": [1, 5, 0, 7, 2],
    "Weather": ["Clear", "Rain", "Clear", "Rain", "Cloudy"],
    "Transport": ["Bus", "Bus", "Car", "Bus", "Bike"],
    "Late": ["No", "Yes", "No", "Yes", "No"]
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
rows, columns= df.shape
print(f"Number of Rows    : {rows}")
print(f"Number of Columns : {columns}")
```

```python
print("Columns:")
for col in df.columns:
    print(col)
```

```python
features = [
    "Distance_KM",
    "Travel_Time_Min",
    "Previous_Late_Count",
    "Weather",
    "Transport"
]

target = "Late"

print("Features:")
for feature in features:
    print(feature)

print("\nTarget:")
print(target)
```

```python
late_counts=df["Late"].value_counts()
print(late_counts.to_string())

# late_employees = (df["Late"] == "Yes").sum()
# not_late_employees = (df["Late"] == "No").sum()

# print("Late Employees     :", late_employees)
# print("Not Late Employees :", not_late_employees)
```

```python
late=df[df["Late"]=="Yes"]["Travel_Time_Min"].mean()
not_late = df[df["Late"]=="No"]["Travel_Time_Min"].mean()

print("Late Employees:")
print(f"{late} minutes")

print("Not Late Employees:")
print(f"{not_late} minutes")
```

```python
df["Prediction"]=((df["Distance_KM"]>15) & (df["Travel_Time_Min"]>60))
df["Prediction"]=df["Prediction"].map({
    True:"Yes",
    False:"No"
})
print(df[["Employee_ID","Late","Prediction"]].to_string(index=False))
```

```python
total_pred=len(df)
cor_pred=(df["Late"]==df["Prediction"]).sum()
acc=cor_pred/total_pred*100

print(f"Correct Predictions : {cor_pred}")
print(f"Total Predictions   : {total_pred}")

print(f"Rule-Based Accuracy : {acc}%")
```

```python
# ML approach instead of rule based 

import pandas as pd
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
data = {
    "Employee_ID": ["E101", "E102", "E103", "E104", "E105"],
    "Distance_KM": [5, 18, 7, 22, 10],
    "Travel_Time_Min": [20, 65, 30, 75, 40],
    "Previous_Late_Count": [1, 5, 0, 7, 2],
    "Weather": ["Clear", "Rain", "Clear", "Rain", "Cloudy"],
    "Transport": ["Bus", "Bus", "Car", "Bus", "Bike"],
    "Late": ["No", "Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

features = [
    "Distance_KM",
    "Travel_Time_Min",
    "Previous_Late_Count",
    "Weather",
    "Transport"
]

target = "Late"
print(df)
```

```python
X,y=df[features],df[target]
print(X)
```

```python
# one hot encode categorical features
categorical_features=[
    "Weather",
    "Transport"
]

print(X[categorical_features])
encoder=OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoded_data=encoder.fit_transform(
    X[categorical_features]
)

encoded_columns=encoder.get_feature_names_out(
    categorical_features
)

encoded_df=pd.DataFrame(
    encoded_data,
    columns=encoded_columns,
    index=X.index
)

print(encoded_df)
```

```python
X_final=pd.concat([X[[
                "Distance_KM",
                "Travel_Time_Min",
                "Previous_Late_Count"
            ]],encoded_df],axis=1)

print(X_final)
```

```python
target_encoder=LabelEncoder()
y_final=target_encoder.fit_transform(y)
print(y_final)
```

```python
model=DecisionTreeClassifier(random_state=42)
model.fit(X_final,y_final)
```

```python
new_employee_1 = pd.DataFrame({
    "Distance_KM": [20],
    "Travel_Time_Min": [70],
    "Previous_Late_Count": [3],
    "Weather": ["Rain"],
    "Transport": ["Bus"]
})
```

```python
new_encoded_1 = encoder.transform(
    new_employee_1[categorical_features]
)

new_encoded_df_1 = pd.DataFrame(
    new_encoded_1,
    columns=encoded_columns
)
print(new_encoded_df_1)
```

```python
new_X_1=pd.concat([new_employee_1[["Distance_KM",
                "Travel_Time_Min",
                "Previous_Late_Count"]],
                 new_encoded_df_1],axis=1)
print(new_X_1)
```

```python
prediction_1=model.predict(new_X_1)
prediction_1 = target_encoder.inverse_transform(
    prediction_1
)

print("\nNew Employee 1")
print("Prediction:", prediction_1[0])
```

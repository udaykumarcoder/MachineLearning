```python
feature_columns = [
    'Area_SqFt',
    'Bedrooms',
    'Building_Age_Yrs',
    'Distance_City_KM',
    'School_Rating',
    'Crime_Index',
    'Interest_Rate_Pct',
    'Property_Tax_Rate'
]

X = df_integrated[feature_columns]

y = df_integrated['Price_USD_M']
```

```python
# Pearson correlation coefficient
numeric_columns=X.columns.tolist()+['Price_USD_M']
# print(numeric_columns) 
# getting list of col names

correlation_matrix=df_integrated[numeric_columns].corr(
    method='pearson'
)
# so now it gets all pair correlation values matrix
price_correlation=correlation_matrix['Price_USD_M'].drop('Price_USD_M')

print("Pearson Correlation with Price_USD_M:\n")

for feature, correlation in price_correlation.items():
    print(f"{feature:<22}: {correlation:+.3f}")
    
```

```python
strongest_positive_feature = price_correlation.idxmax()
strongest_positive_value = price_correlation.max()

strongest_negative_feature = price_correlation.idxmin()
strongest_negative_value = price_correlation.min()

print("\nStrongest Positive Feature:",
      strongest_positive_feature,
      f"({strongest_positive_value:+.3f})")

print("Strongest Negative Feature:",
      strongest_negative_feature,
      f"({strongest_negative_value:+.3f})")
```

```python
high_value=df_integrated[df_integrated['Price_USD_M']>1.5]
moderate_value=df_integrated[df_integrated['Price_USD_M']<=1.5]
```

```python
df_integrated['Value_Group'] = np.where(
    df_integrated['Price_USD_M'] > 1.5,
    'High Value',
    'Moderate Value'
)
```

```python
print(df_integrated.head())
```

```python
group_metrics = df_integrated.groupby('Value_Group').agg(
    Total_Properties=('Property_ID', 'count'),
    Mean_Area=('Area_SqFt', 'mean'),
    Mean_Distance=('Distance_City_KM', 'mean'),
    Mean_School_Rating=('School_Rating', 'mean')
)

print(group_metrics.round(2))
```

```python
master_df=(
    df_demo.merge(df_vitals,on="Patient_ID",how="inner")
    .merge(df_diag,on="Patient_ID",how="inner")
)
```

```python
print(master_df.head())
```

```python
c=master_df["Has_Condition"].value_counts()
print(c)
c_percentage=master_df["Has_Condition"].value_counts(normalize=True)*100
print(c_percentage)
```

```python
print("Class Distribution (Has_Condition):")
print(f"- Negative (0) : {c[0]} patients ({c_percentage[0]:.1f}%)")
print(f"- Positive (1) : {c[1]} patients ({c_percentage[1]}%)")
```

```python
X=master_df.drop(
    columns=["Patient_ID","Has_Condition"]
)
y=master_df["Has_Condition"]
```

```python
# non stratified random train test split
X_train_random,X_test_random, y_train_random, y_test_random=train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
train_random_counts= y_train_random.value_counts()
print(train_random_counts)
train_random_percentage=y_train_random.value_counts(normalize=True)*100
print(train_random_percentage)
```

```python
print("\nTraining Set Size:", len(X_train_random))
print(
    f"Negative (0) : {train_random_counts[0]} "
    f"({train_random_percentage[0]:.1f}%)"
)

print(
    f"Positive (1) : {train_random_counts[1]} "
    f"({train_random_percentage[1]:.1f}%)"
)
```

```python

print("\nTesting Set Size:", len(y_test_random))

test_random_counts = y_test_random.value_counts().sort_index()
test_random_percent = (
    y_test_random.value_counts(normalize=True)
    .sort_index() * 100
)

print(
    f"Negative (0) : {test_random_counts[0]} "
    f"({test_random_percent[0]:.1f}%)"
)

print(
    f"Positive (1) : {test_random_counts[1]} "
    f"({test_random_percent[1]:.1f}%)"
)
```

```python
# calculate distribution drift
full_positive_percentage=(
    y.mean()*100
)
random_test_positive_percentage=(
    y_test_random.mean()*100
)

random_drift=(
    random_test_positive_percentage-full_positive_percentage
)


print("\nDrift Analysis:")
print(
    f"Full Dataset Positive Class : "
    f"{full_positive_percentage:.1f}%"
)

print(
    f"Random Test Positive Class : "
    f"{random_test_positive_percentage:.1f}%"
)

print(
    f"Distribution Drift : "
    f"{random_drift:+.1f} percentage points"
)
```

```python
X_train_strat, X_test_strat, y_train_strat, y_test_strat = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("\nTraining Set Size:", len(y_train_strat))

train_strat_counts = y_train_strat.value_counts().sort_index()
train_strat_percent = (
    y_train_strat.value_counts(normalize=True)
    .sort_index() * 100
)

print(
    f"Negative (0) : {train_strat_counts[0]} "
    f"({train_strat_percent[0]:.1f}%)"
)

print(
    f"Positive (1) : {train_strat_counts[1]} "
    f"({train_strat_percent[1]:.1f}%)"
)


print("\nTesting Set Size:", len(y_test_strat))

test_strat_counts = y_test_strat.value_counts().sort_index()
test_strat_percent = (
    y_test_strat.value_counts(normalize=True)
    .sort_index() * 100
)

print(
    f"Negative (0) : {test_strat_counts[0]} "
    f"({test_strat_percent[0]:.1f}%)"
)

print(
    f"Positive (1) : {test_strat_counts[1]} "
    f"({test_strat_percent[1]:.1f}%)"
)


# Calculate stratified drift

strat_test_positive_percentage = (
    y_test_strat.mean() * 100
)

strat_drift = (
    strat_test_positive_percentage
    - full_positive_percentage
)

print("\nDrift Analysis:")

print(
    f"Full Dataset Positive Class : "
    f"{full_positive_percentage:.1f}%"
)

print(
    f"Stratified Test Positive Class : "
    f"{strat_test_positive_percentage:.1f}%"
)

print(
    f"Distribution Drift : "
    f"{strat_drift:+.1f} percentage points"
)
```

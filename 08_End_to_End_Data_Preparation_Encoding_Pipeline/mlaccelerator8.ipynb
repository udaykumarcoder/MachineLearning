```python
# 1. Problem: What is an outlier?

# An outlier is a value that is unusually far away from most of the other values.

# Your spending data looks roughly like:

# 150
# 170
# 180
# 190
# 210
# ...
# 4,500
# 4,700
# 4,900
# 5,100
# 5,300
# ...
# 11,800
# 12,100
# 12,500
# 12,800
# 13,200
# 13,500
# 14,000
# 15,000

# Most customers are around lower spending levels, while some customers have very high spending.

# We want to identify unusually high values and cap them.

# 2. What is Q1?

# Q1 means:

# 25th percentile

# It is the value below which approximately 25% of the observations fall.

# For your dataset:

# Q1 = $860

# So we can think of:

# 25% of values
#       ↓
# ----- $860 -----------------------------
# Q1

# Values below/around $860 are in the lower quarter of the data.

# 3. What is Q3?

# Q3 means:

# 75th percentile

# For your data:

# Q3 = $5,525

# So:

# ---------------- $5,525 ----------------
#                   ↑
#                   Q3

# Approximately 75% of observations are at or below this value.

# 4. What is IQR?

# IQR means:

# Interquartile Range

# It measures the spread of the middle 50% of your data.

# Formula:

# IQR = Q3 - Q1

# Your values:

# Q3 = 5,525
# Q1 =   860

# Therefore:

# IQR = 5,525 - 860
#     = 4,665

# So:

# IQR = $4,665
# 5. Why multiply by 1.5?

# The standard IQR outlier rule says:

# Upper Bound = Q3 + 1.5 × IQR

# Why?

# We're defining a reasonable upper range around the middle 50% of the data.

# Your calculation:

# Upper Bound
# = 5,525 + 1.5 × 4,665

# First:

# 1.5 × 4,665 = 6,997.50

# Then:

# 5,525 + 6,997.50
# = 12,522.50

# Therefore:

# Upper Bound = $12,522.50
# 6. Now identify outliers

# The rule is:

# If Total_Spend_USD > $12,522.50
#         ↓
#      Outlier

# So:

# 12,000       → Normal
# 12,500       → Normal
# 12,522.50    → Normal
# 12,800       → Outlier
# 13,200       → Outlier
# 13,500       → Outlier
# 14,000       → Outlier
# 15,000       → Outlier
# 7. What does "capping" mean?

# Suppose we have:

# Customer A → $15,000

# The upper limit is:

# $12,522.50

# Instead of deleting the customer, we replace the extreme value with the upper limit:

# Before:

# $15,000

# becomes:

# After:

# $12,522.50

# That's called IQR capping or Winsorization.

# So:

# $13,200 → $12,522.50
# $14,000 → $12,522.50
# $15,000 → $12,522.50

# The customer remains in the dataset; only the extreme numerical value is limited
```

```python
# outlier detection and iqr capping
Q1=df["Total_Spend_USD"].quantile(0.25)
Q3=df["Total_Spend_USD"].quantile(0.75)

IQR=Q3-Q1

lower_bound=Q1-1.5*Q3
upper_bound=Q1+1.5*Q3

outliers=df["Total_Spend_USD"]>upper_bound
outlier_count=outliers.sum()

df["Total_Spend_USD"]=df["Total_Spend_USD"].clip(
    upper=upper_bound
)
```

```python

print("\nIQR Thresholds for Total_Spend_USD:")
print(f"Q1: ${Q1:,.2f}")
print(f"Q3: ${Q3:,.2f}")
print(f"IQR: ${IQR:,.2f}")
print(f"Lower Bound: ${lower_bound:,.2f}")
print(f"Upper Bound: ${upper_bound:,.2f}")
print(f"Outliers Detected: {outlier_count}")
```

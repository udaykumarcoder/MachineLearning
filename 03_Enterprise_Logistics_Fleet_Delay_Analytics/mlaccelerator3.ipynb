```python
df_integrated=df_shipments.merge(df_fleet,on='Vehicle_ID',how='left')
```

```python
df_integrated=df_integrated.merge(df_weather,on='Region_ID',how='left')
```

```python
target = 'Is_Delayed'

drop_columns = [
    'Shipment_ID',
    'Vehicle_ID',
    'Region_ID',
    'Is_Delayed'
]

X=df_integrated.drop(columns=drop_columns)
y=df_integrated[target]
```

```python
numerical_features=X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features=X.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Numerical features")
print(numerical_features)

print("Categorical features")
print(categorical_features)
```

```python
group_metrics=df_integrated.groupby('Is_Delayed')[['Distance_KM','Maintenance_Score']].mean()
print(group_metrics.round(2))
```

```python
condition1=(
    (df_integrated['Distance_KM']>400) & (df_integrated['Maintenance_Score']<60)
)
```

```python
condition2=(
    df_integrated['Weather_Condition'].isin( ['Storm', 'Heavy Rain'])
)
```

```python
rule=condition1 | condition2
```

```python
df_integrated['Rule_Prediction']=np.where(
    rule,
    'Yes',
    'No'
)
```

```python
df_integrated[
    [
        'Shipment_ID',
        'Distance_KM',
        'Maintenance_Score',
        'Weather_Condition',
        'Rule_Prediction'
    ]
].head(10)
```

```python
correct_predictions = (
    df_integrated['Rule_Prediction'] ==
    df_integrated['Is_Delayed']
).sum()

print("Correct Predictions:", correct_predictions)
```

```python
accuracy=(df_integrated['Rule_Prediction']==df_integrated['Is_Delayed']).mean()*100
print(f"Accuracy :{accuracy:.2f}%")
```

```python
false_positive_cases=df_integrated[(df_integrated['Is_Delayed']=='No') & (df_integrated['Rule_Prediction']=='Yes')]
false_positive_cases[
    [
        'Shipment_ID',
        'Distance_KM',
        'Maintenance_Score',
        'Weather_Condition',
        'Traffic_Congestion_Index',
        'Is_Delayed',
        'Rule_Prediction'
    ]
]
```

```python
print("=" * 60)
print("      ENTERPRISE LOGISTICS DELAY ANALYTICS")
print("=" * 60)

# Dataset information
print(f"\nIntegrated Dataset Records : {len(df_integrated)}")
print(f"Total Combined Columns    : {df_integrated.shape[1]}")
print(f"Predictor Features        : {X.shape[1]}")

# Target distribution
no_count = (df_integrated['Is_Delayed'] == 'No').sum()
yes_count = (df_integrated['Is_Delayed'] == 'Yes').sum()

no_pct = no_count / len(df_integrated) * 100
yes_pct = yes_count / len(df_integrated) * 100

print(
    f"Target Class Distribution : "
    f"No = {no_count} ({no_pct:.1f}%), "
    f"Yes = {yes_count} ({yes_pct:.1f}%)"
)

# Group metrics
metrics = df_integrated.groupby('Is_Delayed')[
    ['Distance_KM', 'Maintenance_Score']
].mean()

print("\nGroup Metrics (Delayed vs On-Time):")

print(
    f"- Mean Distance            : "
    f"Delayed = {metrics.loc['Yes', 'Distance_KM']:.2f} KM | "
    f"On-Time = {metrics.loc['No', 'Distance_KM']:.2f} KM"
)

print(
    f"- Mean Vehicle Maintenance : "
    f"Delayed = {metrics.loc['Yes', 'Maintenance_Score']:.2f} Score | "
    f"On-Time = {metrics.loc['No', 'Maintenance_Score']:.2f} Score"
)

# Rule evaluation
correct = (
    df_integrated['Rule_Prediction'] ==
    df_integrated['Is_Delayed']
).sum()

accuracy = correct / len(df_integrated) * 100

fp = (
    (df_integrated['Is_Delayed'] == 'No') &
    (df_integrated['Rule_Prediction'] == 'Yes')
).sum()

fn = (
    (df_integrated['Is_Delayed'] == 'Yes') &
    (df_integrated['Rule_Prediction'] == 'No')
).sum()

print("\nRule Engine Evaluation:")
print(f"- Rule Accuracy           : {accuracy:.2f}%")
print(f"- Correct Predictions     : {correct}")
print(f"- False Positives         : {fp}")
print(f"- False Negatives         : {fn}")

print("\nConclusion:")
print(
    "Multi-source integration combines shipment, fleet, and weather "
    "information into a single analytical dataset. The supplied rule "
    "achieves 64% accuracy, with 10 false positives and 8 false negatives. "
    "The failures demonstrate that fixed thresholds cannot fully capture "
    "the interactions among distance, vehicle condition, congestion, "
    "weather, cargo, and driver-related factors. These integrated features "
    "can therefore serve as the foundation for a future machine-learning "
    "classification model."
)
```

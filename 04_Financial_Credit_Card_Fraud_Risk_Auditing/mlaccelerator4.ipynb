```python
# ============================================================
# FINANCIAL CREDIT CARD FRAUD & RISK AUDITING
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# TASK 1 — 4-TABLE RELATIONAL MERGE
# ============================================================

df_master = df_tx.merge(
    df_acc,
    on='Account_ID',
    how='left'
)

df_master = df_master.merge(
    df_trm,
    on='Terminal_ID',
    how='left'
)

df_master = df_master.merge(
    df_blk,
    on='Tx_ID',
    how='left'
)

print("========== MASTER DATASET ==========")
print("Rows:", df_master.shape[0])
print("Columns:", df_master.shape[1])


# ============================================================
# TASK 2 — SCHEMA VALIDATION
# ============================================================

print("\n========== SCHEMA VALIDATION ==========")

print("\nDataset Dimensions:")
print(df_master.shape)

print("\nData Types:")
print(df_master.dtypes)

print("\nMissing Values:")
print(df_master.isnull().sum())

print("\nMissing Value Percentage:")
print((df_master.isnull().mean() * 100).round(2))


# ============================================================
# TASK 3 — FEATURE SPACE
# ============================================================

operational_ids = [
    'Tx_ID',
    'Account_ID',
    'Terminal_ID'
]

feature_columns = [
    'Tx_Amount',
    'Tx_Hour',
    'Avg_Monthly_Tx_Vol',
    'Account_Age_Months',
    'Terminal_Risk_Score',
    'Is_Foreign_Location',
    'High_Risk_IP_Flag'
]

target = 'Is_Fraud'

X = df_master[feature_columns]
y = df_master[target]

print("\n========== FEATURE SPACE ==========")

print("Operational IDs:")
print(operational_ids)

print("\nPredictive Features:")
print(feature_columns)

print("\nTarget:")
print(target)


# ============================================================
# TASK 4 — FRAUD CLASS AGGREGATION
# ============================================================

print("\n========== CLASS DISTRIBUTION ==========")

class_counts = df_master['Is_Fraud'].value_counts().sort_index()

class_percentages = (
    df_master['Is_Fraud']
    .value_counts(normalize=True)
    .sort_index() * 100
)

print("Counts:")
print(class_counts)

print("\nPercentages:")
print(class_percentages.round(2))

print("\nGroup Medians:")

group_medians = df_master.groupby('Is_Fraud')[
    ['Tx_Amount', 'Terminal_Risk_Score']
].median()

print(group_medians)


# ============================================================
# TASK 5 — RULE-BASED FRAUD SYSTEM
# ============================================================

amount_hour_rule = (
    (df_master['Tx_Amount'] > 1000) &
    (df_master['Tx_Hour'].isin([1, 2, 3, 4]))
)

ip_rule = (
    df_master['High_Risk_IP_Flag'] == 1
)

df_master['Rule_Prediction'] = (
    amount_hour_rule | ip_rule
).astype(int)


print("\n========== RULE PREDICTIONS ==========")

print(
    df_master[
        [
            'Tx_ID',
            'Tx_Amount',
            'Tx_Hour',
            'High_Risk_IP_Flag',
            'Is_Fraud',
            'Rule_Prediction'
        ]
    ]
)


# ============================================================
# TASK 6 — PERFORMANCE
# ============================================================

actual = df_master['Is_Fraud']
predicted = df_master['Rule_Prediction']

true_positives = (
    (actual == 1) &
    (predicted == 1)
).sum()

true_negatives = (
    (actual == 0) &
    (predicted == 0)
).sum()

false_positives = (
    (actual == 0) &
    (predicted == 1)
).sum()

false_negatives = (
    (actual == 1) &
    (predicted == 0)
).sum()

correct_predictions = (
    actual == predicted
).sum()

accuracy = correct_predictions / len(df_master)

print("\n========== RULE PERFORMANCE ==========")

print("Correct Detections:", true_positives)
print("Correct Legitimate:", true_negatives)
print("False Alarms (FP):", false_positives)
print("Missed Frauds (FN):", false_negatives)
print("Accuracy:", round(accuracy * 100, 2), "%")


# ============================================================
# TASK 7 — TRAIN VS LIVE PREDICTION
# ============================================================

print("\n========== TRAIN VS PREDICT ==========")

# Historical training data
X_train = df_master[feature_columns]
y_train = df_master['Is_Fraud']

print("Training X shape:", X_train.shape)
print("Training y shape:", y_train.shape)

# Simulated live transactions
live_transactions = df_master[
    feature_columns
].head(5).copy()

print("\nLive Transaction Stream:")
print(live_transactions)

print("\nTraining:")
print("X + y are available.")

print("\nLive Prediction:")
print("Only X is available.")
```

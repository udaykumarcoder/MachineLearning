```python
df_integrated=(
    df_activity.merge(df_sub,on='User_ID',how='inner')
    .merge(df_device,on='Device_ID',how='inner')
)
print(df_integrated.head())
```

```python

print("Total Combined Records  :", len(df_integrated))
print("Total Combined Columns  :", len(df_integrated.columns))
# Check whether a target column exists
target_columns = [
    'Target',
    'target',
    'Label',
    'label',
    'y'
]

existing_targets = [
    col for col in target_columns
    if col in df_integrated.columns
]


if len(existing_targets) == 0:
    y = None
    print("Target Vector (y)      : ABSENT (Unsupervised Setup)")
    print("Paradigm                : Unsupervised Learning / Clustering")
else:
    y = df_integrated[existing_targets[0]]
    print("Target Vector (y)      : PRESENT")
    print("Paradigm                : Supervised Learning")

print("\nColumn Data Types:")
print(df_integrated.dtypes)
```

```python
# Select numeric columns
numeric_columns = df_integrated.select_dtypes(
    include=np.number
).columns.tolist()
print(numeric_columns)
```

```python
feature_columns = [
    'Daily_Listening_Mins',
    'Playlists_Created',
    'Skip_Ratio',
    'Podcast_Listening_Pct',
    'Account_Age_Months'
]


X = df_integrated[feature_columns]
```

```python
print("\nNumeric Features (X) Range Analysis:")

for column in feature_columns:

    minimum = X[column].min()
    maximum = X[column].max()
    span = maximum - minimum

    print(
        f"{column:25s} : "
        f"Min = {minimum}, "
        f"Max = {maximum}, "
        f"Span = {span}"
    )
```

```python

print("\nFeature Variance:")
print(X.var())
```

```python

print("\nScale Analysis:")
print(
    "Daily_Listening_Mins has a much larger numerical scale "
    "than ratio-based features."
)

print(
    "Therefore, scaling is required before distance-based "
    "clustering algorithms such as K-Means."
)
```

```python
df_integrated['Heuristic_Cluster']=np.where(
    (df_integrated['Daily_Listening_Mins']>100)
    &
    (df_integrated['Monthly_Fee_USD']>0),
    0,
    1
)
```

```python
cluster_counts=df_integrated['Heuristic_Cluster'].value_counts()
print(cluster_counts)
```

```python
for cluster,count in cluster_counts.items():
    percentage=count/len(df_integrated)*100
    if cluster == 0:
        name = "Power Listeners"
    else:
        name = "Casual Listeners"
    print(
        f"Cluster {cluster} ({name}) : "
        f"{count} users ({percentage:.1f}%)"
    )
```

```python
print(df_integrated.head())
```

```python
cluster_profile=df_integrated.groupby('Heuristic_Cluster')[feature_columns].mean()
print(cluster_profile)
```

```python
print("\nMean Profile by Cluster:")

print(
    cluster_profile.to_string(
        float_format=lambda x: f"{x:.3f}"
    )
)
```

```python
print("\nMean Profile by Cluster:")

for cluster in cluster_profile.index:

    if cluster == 0:
        name = "Power"
    else:
        name = "Casual"

    print(
        f"Cluster {cluster} ({name})",
        end=" "
    )

    for feature in feature_columns:
        print(
            f"{cluster_profile.loc[cluster, feature]:.3f}",
            end=" "
        )

    print()
```
